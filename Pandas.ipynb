```python
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya", "Simran", "Karan"],
    "Age": [21, 22, 20, 23, 21],
    "Salary": [30000, 45000, 35000, 50000, 40000],
    "Department": ["IT", "HR", "IT", "Finance", "HR"]
}

df = pd.DataFrame(data)
```

```python
#Selecting Rows and Columns in Pandas
#only 1 column
print(df["Name"])
```

```python
#Multiple columns
print(df[["Name","Age","Salary"]])
```

```python
#row
print(df.iloc[0])
```

```python
#multiple rows
print(df.iloc[0:3])
```

```python
#for particular position
print(df.iloc[0,3])
```

```python
#Select row using loc
print(df.loc[0:2, ["Name","Salary"]])
```

```python
#Filtering row using conditions
result = df[df["Salary"] > 40000]
print(result)
```

```python
#filtering using multiple conditions
result = df[
  (df["Age"] > 30) &
  (df["Salary"] > 30000)
]
print(result)
```

```python
#filter using OR condition
result = df[
  (df["Department"] == "HR") |
  (df["Department"] == "IT")
]
print(result)
```

```python
#Filter using isin()
result = df[
  (df["Department"].isin(["IT"])) 
]
print(result)
```

```python
#Filter using between()
result = df[df["Salary"].between(30000,50000)]
print(result)
```

```python
#Adding a new column
df["Bonus"] = df["Salary"] * 2
print(df)
```

```python
#Delete a column
df = df.drop(columns=["Age"])
print(df)
```

```python
#Updating an existing column
df["Salary"] = df["Salary"] + 10000
print(df)
```

```python
#Renaming a column
df = df.rename(columns={"Salary":"Monthly_salary"})
print(df)
```

```python
#sorting data in pandas
df = df.sort_values(by="Monthly_salary",ascending=False)
print(df)
```

```python
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya", "Simran", "Karan"],
    "Age": [21, None, 20, 23, None],
    "Salary": [30000, 45000, None, 50000, 40000],
    "Department": ["IT", "HR", None, "Finance", "HR"]
}

df = pd.DataFrame(data)
```

```python
#Handling missing values in pandas
#Detecting Missing Value
df.isnull()
```

```python
#To count missing value in each column
df.isnull().sum()
```

```python
#check whether the Entire dataset has missing value
df.isnull().sum().sum()
```

```python
#Remove missing value
df.dropna()
```

```python
# fill missing value 
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print(df)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
print(df)
```

```python
#Handling Duplicates in pandas
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Amit", "Raj", "Karan"],
    "Age": [21, 23, 21, 23, 24],
    "Salary": [30000, 45000, 30000, 50000, 40000]
}

df = pd.DataFrame(data)
```

```python
# Find duplicates
df.duplicated()
```

```python
#count of duplicates
df.duplicated().sum()
```

```python

# remove dupllicates rows
df = df.drop_duplicates()
print(df)
```

```python
# Find duplicates using specific columnn
df.duplicated(subset=["Name"])
```

```python
# Remove duplicates
df = df.drop_duplicates(subset=["Name"])
print(df)
```

```python
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya", "Simran", "Karan"],
    "Age": [21, 22, 20, 23, 21],
    "Salary": [30000, 45000, 35000, 50000, 40000],
    "Department": ["IT", "HR", "IT", "Finance", "HR"]
}

df = pd.DataFrame(data)
```

```python
# counting categories using value_counts()
df["Department"].value_counts()
```

```python
#calculate percentage
df["Department"].value_counts(normalize=True) * 100
```

```python
# Unique categories
df["Department"].unique()
```

```python
# count Unique categories
df["Department"].nunique()
```

```python
# combining dataset
import pandas as pd

jan = pd.DataFrame({
    'Product': ['Mobile', 'Laptop'],
    'Sales': [100, 50]
})

feb = pd.DataFrame({
    'Product': ['Tablet', 'Watch'],
    'Sales': [80, 40]
})

result = pd.concat([jan, feb], axis=0, ignore_index=True)

print(result)
```
