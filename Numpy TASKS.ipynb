```python
#create 2 * 2 matrices and multiply using @
import numpy as np

mat1 = np.array([
  [1,2],
  [4,5]
])

mat2 = np.array([
  [7,8],
  [2,5]
])

result = mat1 @ mat2
print(result)

result2 = mat1 * mat2
print(result2)
```

```python
import numpy as np

# 1. Create dataset
marks = np.array([
    [85, 90, 78, 92],
    [65, 70, 75, 80],
    [95, 88, 92, 96],
    [40, 55, 60, 50],
    [72, 68, 85, 79]
])

# 2. Dataset inspection
print("Shape:", marks.shape)
print("Dimensions:", marks.ndim)
print("Students:", marks.shape[0])
print("Subjects:", marks.shape[1])

# 3. Statistical analysis
subject_average = np.mean(marks, axis=0)
student_average = np.mean(marks, axis=1)

print("\nSubject averages:", subject_average)
print("Student averages:", student_average)

print("Highest marks:", np.max(marks))
print("Lowest marks:", np.min(marks))

# 4. Boolean indexing
print("\nMarks above 80:", marks[marks > 80])

print("Student averages above 75:",
      student_average[student_average > 75])

# 5. Broadcasting
updated_marks = marks + 5

print("\nUpdated marks:\n", updated_marks)

updated_average = np.mean(updated_marks, axis=1)
print("Updated student averages:", updated_average)

# 6. Standard deviation
subject_std = np.std(marks, axis=0)

print("\nSubject standard deviation:", subject_std)

subject_names = np.array([
    "Maths", "Physics", "Chemistry", "Computer"
])

index = np.argmax(subject_std)

print("Greatest variation:", subject_names[index])
```
